# Checking exact duplicate rows

**Before removing the duplicate rows**



```python
print("Before removing the duplicate rows")
print("Movies duplicates:", movies.duplicated().sum())
print("Genres duplicates:", genres.duplicated().sum())
print("movie genres duplicates:", movie_genres.duplicated().sum())
print("cast duplicates:", cast.duplicated().sum())
print("crew duplicates:", crew.duplicated().sum())
print("Movie keywords duplicates:", movie_keywords.duplicated().sum())
```

```python
movies = movies.drop_duplicates()
genres = genres.drop_duplicates()
movie_genres = movie_genres.drop_duplicates()
cast = cast.drop_duplicates()
crew = crew.drop_duplicates()
movie_keywords = movie_keywords.drop_duplicates()
```

**After removing the duplicate rows**

```python
print("After removing the duplicate rows")
print("Movies duplicates:", movies.duplicated().sum())
print("Genres duplicates:", genres.duplicated().sum())
print("Movie genres duplicates:", movie_genres.duplicated().sum())
print("Cast duplicates:", cast.duplicated().sum())
print("Crew duplicates:", crew.duplicated().sum())
print("Movie keywords duplicates:", movie_keywords.duplicated().sum())
```

# Check duplicate movie IDs

```python
movies["movie_id"].nunique()
```

**Check movie IDs that do not exist in movies.csv**

```python
movie_ids = movies["movie_id"].unique()

print("Orphan movie_genres rows:",
      (~movie_genres["movie_id"].isin(movie_ids)).sum())

print("Orphan cast rows:",
      (~cast["movie_id"].isin(movie_ids)).sum())

print("Orphan crew rows:",
      (~crew["movie_id"].isin(movie_ids)).sum())

print("Orphan movie_keywords rows:",
      (~movie_keywords["movie_id"].isin(movie_ids)).sum())
```

```python
movie_genres = movie_genres[movie_genres["movie_id"].isin(movie_ids)]
cast = cast[cast["movie_id"].isin(movie_ids)]
crew = crew[crew["movie_id"].isin(movie_ids)]
movie_keywords = movie_keywords[movie_keywords["movie_id"].isin(movie_ids)]
```

# Save the cleaned CSV files

```python
from pathlib import Path

output_folder = Path("/content/drive/MyDrive/Cleaned data")

movies.to_csv("/content/drive/MyDrive/Cleaned data/ movies.csv", index=False)
genres.to_csv("/content/drive/MyDrive/Cleaned data/ genres.csv", index=False)
movie_genres.to_csv("/content/drive/MyDrive/Cleaned data/ movie_genres.csv", index=False)
cast.to_csv("/content/drive/MyDrive/Cleaned data/ cast.csv", index=False)
crew.to_csv("/content/drive/MyDrive/Cleaned data/ crew.csv", index=False)
movie_keywords.to_csv("/content/drive/MyDrive/Cleaned data/ movie_keywords.csv", index=False)

print("Cleaned CSV files saved in:", output_folder.resolve())
```
